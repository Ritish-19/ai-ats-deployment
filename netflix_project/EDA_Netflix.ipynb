```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
NETFLIX_RED = "#E50914"
BLUE        = "#2E75B6"
DARK        = "#221F1F"
ORANGE      = "#E6920A"
LIGHT_GRAY  = "#F5F5F5"
plt.rcParams.update({
    "figure.facecolor"  : "white",
    "axes.facecolor"    : LIGHT_GRAY,
    "axes.spines.top"   : False,
    "axes.spines.right" : False,
    "axes.grid"         : True,
    "axes.grid.axis"    : "x",
    "grid.color"        : "#DDDDDD",
    "grid.linewidth"    : 0.6,
    "font.size"         : 11,
})
```

```python
# ── LOAD DATA ────────────────────────────────────────────────
print("Loading cleaned data ...")
df = pd.read_csv("cleaned_netflix.csv", parse_dates=["date_added"])
g  = pd.read_csv("genres_exploded.csv")
c  = pd.read_csv("countries_exploded.csv")
print(f"  Main: {len(df):,}   Genres: {len(g):,}   Countries: {len(c):,}")
```

```python
# ════════════════════════════════════════════════════════════
# CHART 1 — Movies vs TV Shows (Donut)
# Library: matplotlib.pyplot
# ════════════════════════════════════════════════════════════
print("\nChart 1: Movies vs TV Shows ...")
counts = df["type"].value_counts()
 
fig, ax = plt.subplots(figsize=(7, 7), facecolor="white")
wedges, texts, autotexts = ax.pie(
    counts, labels=counts.index,
    autopct="%1.1f%%", startangle=90,
    colors=[NETFLIX_RED, BLUE],
    wedgeprops={"width": 0.55, "edgecolor": "white", "linewidth": 3},
    textprops={"fontsize": 13},
)
for at in autotexts:
    at.set_fontsize(14); at.set_fontweight("bold"); at.set_color("white")
ax.set_title("Movies vs TV Shows on Netflix", fontsize=16, fontweight="bold", pad=20)
ax.text(0, 0, f"{len(df):,}\nTitles", ha="center", va="center",
        fontsize=14, fontweight="bold", color=DARK)
ax.set_facecolor("white")
plt.tight_layout()
plt.savefig("chart1_movies_vs_tvshows.png", dpi=150, bbox_inches="tight")
plt.close()
print("  Saved chart1_movies_vs_tvshows.png")
 
```
