```python
import numpy as np
import pandas as pd
```

```python
#load the data
df=pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum() #finding the null value.
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda a:a.fillna(a.median()))
df.info() 
```

```python
#checking the null values
df.isnull().sum()
```

```python
#converting the columns names into appropritae way
#To make our code consistent and easy to read and in snake chasing it is easy
#to refrebnce our code in sql or python without getting error

df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.head()
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns  ##rename  the column.
```

```python
df.head()
```

```python
#To check the unique values to modify it
df['frequency_of_purchases'].unique()
```

```python
# Modifying the values
a={'Fortnightly':14,'Weekly':7,'Annually':365,'Quarterly':90,'Bi-weekly':14,'Monthly':30,'Every 3 Months':90}
df['frequency_of_purchases']=df['frequency_of_purchases'].map(a)
df
```
